```python
import nltk
import re

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    text = text.lower()

    # keep letters and spaces
    text = re.sub(r'[^a-z\s]', '', text)

    tokens = text.split()

    filtered_tokens = []
    for w in tokens:
        if w not in stop_words:
            filtered_tokens.append(w)

    lemmatized_tokens = []
    for w in filtered_tokens:
        lemmatized_tokens.append(lemmatizer.lemmatize(w))

    return ' '.join(lemmatized_tokens)

print(preprocess('I am going to shopping with my friends'))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
emails=[
    'cricket match cricket match cricket',
    'free win money free win money',
    'cricket match free win money cricket'
]
cv=CountVectorizer()
X=cv.fit_transform(emails).toarray()
print('vocab:',cv.get_feature_names_out())
print('Bow matrix:\n',X)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
emails=[
    'cricket match cricket match cricket',
    'free win money free win money',
    'cricket match free win money cricket'
]
tfidf=TfidfVectorizer(sublinear_tf=True,max_df=0.95,min_df=1)
X=tfidf.fit_transform(emails).toarray()
print('Feature names:',tfidf.get_feature_names_out())
print('Tfidf matrix:\n',X)
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

def preprocess(text):

    text = text.lower()

    text = re.sub(r'[^a-z\s]', '', text)


    tokens = word_tokenize(text)


    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]


    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    return tokens

text = "Hello!!! My name is Prem. I am learning NLP, and it's AMAZING!!! 😊😊 Running, runs, ran..."
print(preprocess(text))
```
