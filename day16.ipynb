```python
import nltk
import re

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    text = text.lower()

    # keep letters and spaces
    text = re.sub(r'[^a-z\s]', '', text)

    tokens = text.split()

    filtered_tokens = []
    for w in tokens:
        if w not in stop_words:
            filtered_tokens.append(w)

    lemmatized_tokens = []
    for w in filtered_tokens:
        lemmatized_tokens.append(lemmatizer.lemmatize(w))

    return ' '.join(lemmatized_tokens)

print(preprocess('I am going to shopping with my friends'))
```

```python
import re
import nltk

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')

nltk.download('punkt_tab')
nltk.download('omw-1.4')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

text = "Hello!!! My name is nicky. I am learning NLP subject, and it's AMAZING!!! 😊 Running, runs, ran..."

def preprocess(text):

    text = text.lower()


    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    tokens = word_tokenize(text)


    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens
              if word not in stop_words]

    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word, pos='v') for word in tokens]

    return tokens

result = preprocess(text)

print(result)
```

```python
import pandas as pd

data = {
    "I saw the man with the telescope.":
        ("Ambiguity", "Not clear who has the telescope."),

    "Oh wonderful, my internet stopped during the interview.":
        ("Sarcasm", "Positive words express a negative situation."),

    "He kicked the bucket yesterday.":
        ("Context/Idiom", "Means died, not literally kicking a bucket."),

    "BRB, I need to answer a call.":
        ("Abbreviation/Slang", "BRB means 'Be Right Back'."),

    "Chat is my favorite animal.":
        ("Context Ambiguity", "Chat can mean an animal or a conversation.")
}

rows = []

for sentence, (challenge, reason) in data.items():
    rows.append({
        "Sentence": sentence,
        "NLP Challenge": challenge,
        "Reason": reason
    })

df = pd.DataFrame(rows)

print(df)
```

```python
import re
import math
from collections import Counter

docs = [
    "I love cricket",
    "I love football",
    "Cricket is my favorite sport"
]

stopwords = {"i", "is", "my"}
processed_docs = []
for doc in docs:
    doc = doc.lower()
    words = re.findall(r'\b[a-z]+\b', doc)
    words = [w for w in words if w not in stopwords]
    processed_docs.append(words)

print("Processed Documents:")
print(processed_docs)
vocab = sorted(set(word for doc in processed_docs for word in doc))
print("\nVocabulary:")
print(vocab)

print("\nBag of Words:")
bow_matrix = []
for doc in processed_docs:
    counts = Counter(doc)
    row = [counts[word] for word in vocab]
    bow_matrix.append(row)
print("Vocabulary:", vocab)

for i, row in enumerate(bow_matrix, 1):
    print(f"Doc{i}: {row}")

N = len(processed_docs)

idf = {}

for word in vocab:
    df = sum(word in doc for doc in processed_docs)
    idf[word] = math.log(N / df)

print("\nIDF Scores:")

for word, score in idf.items():
    print(word, round(score, 3))

print("\nTF-IDF Scores:")

for i, doc in enumerate(processed_docs, 1):
    counts = Counter(doc)

    print(f"\nDoc{i}")

    for word in vocab:
        tf = counts[word] / len(doc)
        tfidf = tf * idf[word]

        if tfidf > 0:
            print(word, round(tfidf, 3))
```
